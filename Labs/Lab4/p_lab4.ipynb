Навчитися здійснювати маніпуляції з динамічними списками (list), застосовувати адресну індексацію, операції вилучення зрізів (slicing) та декларативні генератори списків (List Comprehensions). **Завдання 1 (Вектори):** генерація вектора довжиною $N=16$ у межах $[-35, 35]$; фільтрація/модифікація через List Comprehension: заміна парних елементів їхніми квадратами; пошук специфічних екстремумів; сортування вставками (Insertion Sort) без використання .sort().

**Завдання 2 (Матриці):** генерація цілочисельної матриці $A$ ($4 \times 5$); спіральний обхід матриці з формуванням одновимірного списку; множення $A \times B$ на сумісну матрицю $B$ ($5 \times L$); пошук усіх сідлових точок (мінімум у рядку і максимум у стовпці або навпаки).

```python
import random

# Завдання 3: Кортеж конфігурації та паралельне розпакування
print("=" * 20, "Завдання 3: Кортеж конфігурації та паралельне розпакування", "=" * 20)
# CONFIG = (N, M, K, X_MIN, X_MAX)
CONFIG = (16, 4, 5, -35, 35)
N, M, K, X_MIN, X_MAX = CONFIG

print(f"Конфігурація: N={N}, M={M}, K={K}, діапазон=[{X_MIN}, {X_MAX}]\n")


# Завдання 1: Вектори
print("=" * 20, "Завдання 1: Вектори", "=" * 20)

# Генерація випадкового вектора V
V = [random.randint(X_MIN, X_MAX) for _ in range(N)]
print(f"Початковий вектор V:\n{V}")

# Модифікація за умовою: заміна парних елементів їхніми квадратами через List Comprehension
# (парні числа: x % 2 == 0)
V_filtered = [x**2 if x % 2 == 0 else x for x in V]
print(f"\nМодифікований вектор (парні замінено квадратами):\n{V_filtered}")

# Знаходження екстремумів (без вбудованого min/max для демонстрації базових алгоритмів)
def find_extremums(arr):
    min_val = arr[0]
    max_val = arr[0]
    for val in arr[1:]:
        if val < min_val:
            min_val = val
        if val > max_val:
            max_val = val
    return min_val, max_val

min_elem, max_elem = find_extremums(V_filtered)
print(f"\nЕкстремуми: Мінімальний = {min_elem}, Максимальний = {max_elem}")

# Сортування вставками (Insertion Sort) без .sort()
def insertion_sort(arr):
    sorted_arr = arr.copy()
    for i in range(1, len(sorted_arr)):
        key = sorted_arr[i]
        j = i - 1
        while j >= 0 and sorted_arr[j] > key:
            sorted_arr[j + 1] = sorted_arr[j]
            j -= 1
        sorted_arr[j + 1] = key
    return sorted_arr

V_sorted = insertion_sort(V_filtered)
print(f"\nВектор після сортування вставками:\n{V_sorted}\n")


# Завдання 2: Матриці
print("=" * 20, "Завдання 2: Матриці", "=" * 20)

# Генерація матриці A розмірності M x K (4 x 5)
A = [[random.randint(X_MIN, X_MAX) for _ in range(K)] for _ in range(M)]

print("Матриця A (4x5):")
for row in A:
    print(" ".join(f"{val:4d}" for val in row))

# Спіральний обхід матриці (за годинниковою стрілкою)
def spiral_order(matrix):
    result = []
    if not matrix:
        return result

    top, bottom = 0, len(matrix) - 1
    left, right = 0, len(matrix[0]) - 1

    while top <= bottom and left <= right:
        # Рух вправо по верхній межі
        for col in range(left, right + 1):
            result.append(matrix[top][col])
        top += 1

        # Рух вниз по правій межі
        for row in range(top, bottom + 1):
            result.append(matrix[row][right])
        right -= 1

        # Рух вліво по нижній межі
        if top <= bottom:
            for col in range(right, left - 1, -1):
                result.append(matrix[bottom][col])
            bottom -= 1

        # Рух вгору по лівій межі
        if left <= right:
            for row in range(bottom, top - 1, -1):
                result.append(matrix[row][left])
            left += 1

    return result

spiral_list = spiral_order(A)
print(f"\nОдновимірний список спірального обходу:\n{spiral_list}")

# Множення на сумісну матрицю B (розмірність K x L, нехай L = 3)
L = 3
B = [[random.randint(-10, 10) for _ in range(L)] for _ in range(K)]

print(f"\nСумісна матриця B (5x{L}):")
for row in B:
    print(" ".join(f"{val:4d}" for val in row))

# Обчислення добутку A x B вручну
def matrix_multiply(mat1, mat2):
    rows1, cols1 = len(mat1), len(mat1[0])
    cols2 = len(mat2[0])
    res = [[0] * cols2 for _ in range(rows1)]

    for r in range(rows1):
        for c in range(cols2):
            for k_idx in range(cols1):
                res[r][c] += mat1[r][k_idx] * mat2[k_idx][c]
    return res

C = matrix_multiply(A, B)
print(f"\nРезультат матричного добутку C = A x B (4x{L}):")
for row in C:
    print(" ".join(f"{val:6d}" for val in row))

# Пошук усіх сідлових точок
# Класичне визначення сідлової точки: елемент одночасно є
# (мінімумом у рядку ТА максимумом у стовпці) АБО (максимумом у рядку ТА мінімумом у стовпці)
def find_saddle_points(matrix):
    rows = len(matrix)
    cols = len(matrix[0])
    saddle_points = []

    row_mins = [min(row) for row in matrix]
    row_maxs = [max(row) for row in matrix]
    col_mins = [min(matrix[r][c] for r in range(rows)) for c in range(cols)]
    col_maxs = [max(matrix[r][c] for r in range(rows)) for c in range(cols)]

    for r in range(rows):
        for c in range(cols):
            val = matrix[r][c]
            # Тип 1: мінімум у рядку і максимум у стовпці
            # Тип 2: максимум у рядку і мінімум у стовпці
            if (val == row_mins[r] and val == col_maxs[c]) or (val == row_maxs[r] and val == col_mins[c]):
                saddle_points.append((r, c, val))

    return saddle_points

saddles = find_saddle_points(A)
print("\nСідлові точки матриці A [індекс рядка, індекс стовпця, значення]:")
if saddles:
    for pt in saddles:
        print(f"Елемент A[{pt[0]}][{pt[1]}] = {pt[2]}")
else:
    print("Сідлових точок у згенерованій матриці не виявлено.")
```

**Контрольні запитання**

1. Структура PyListObject та складність операцій
Внутрішня структура: PyListObject містить заголовок PyVarObject (лічильник посилань, тип, довжина ob_size), ємність виділеного буфера allocated та вказівник ob_item на неперервний динамічний масив вказівників на PyObject*.
Складність $L[i]$: адреса елемента обчислюється миттєво за формулою прямої адресації ob_item + i * sizeof(PyObject*), тому час доступу становить $O(1)$.
Складність $L.\text{insert}(0, x)$: вставка на початок вимагає фізичного зсуву всіх наявних $n$ вказівників у масиві праворуч на одну позицію через memmove, тому складність становить $O(n)$.
2. Спільний стан при [[0] * 4] * 3 та list comprehension
Причина помилки: зовнішній оператор * 3 не клонує списки, а копіює тричі посилання на один і той самий об'єкт рядка [0, 0, 0, 0] у купі; зміна елемента в одному рядку призводить до зміни у всіх трьох.
Генератор списків: вираз [[0] * 4 for _ in range(3)] на кожній ітерації циклу викликає конструктор нового списку, створюючи три повністю незалежні об'єкти списків у пам'яті.
3. Зрізи (slicing) та робота з пам'яттю
Математичний зміст: вибірка підмножини елементів послідовності за арифметичною прогресією індексів $i_k = \text{start} + k \cdot \text{step}$ для $k \ge 0$ при $i_k < \text{stop}$.
Новий масив: операція sub = L[1:5] виділяє новий блок пам'яті під новий PyListObject і копіює туди відповідні вказівники (поверхнева копія).
Вплив: вихідний список $L$ ніяк не змінюється; модифікація структури sub не зачіпає $L$.
4. Сідлова точка в матриці
Математична умова: елемент $A_{ij}$ є сідловою точкою, якщо він є одночасно мінімумом у своєму рядку і максимумом у своєму стовпчику (або навпаки: максимумом у рядку та мінімумом у стовпчику):
   $$A_{ij} = \min_{k} A_{ik} \quad \text{та} \quad A_{ij} = \max_{m} A_{mj}$$
Кількість точок: мінімум — $0$ (наприклад, для циклічно зсунутих значень); максимум — $m \times n$ (для прямокутної матриці, де всі елементи однакові).
5. Сортування вибором (Selection Sort)
Принцип: масив ділиться на відсортовану та невідсортовану частини; на кожному кроці знаходиться мінімальний елемент серед невідсортованих і міняється місцями з першим невідсортованим елементом.
Складність: кількість порівнянь завжди дорівнює $\frac{n(n-1)}{2}$, тому в найкращому, середньому та найгіршому випадках складність становить строго $O(n^2)$.
6. Незмінність tuple та оптимізація CPython
Чому незмінні: значення та довжина кортежу фіксуються під час створення для гарантії цілісності та можливості використання кортежів як ключів хеш-таблиць (словників і множин).
Архітектурні переваги: CPython не виділяє надлишкової пам'яті про запас (відсутнє поле allocated, виділяється рівно стільки пам'яті, скільки потрібно для $n$ елементів); використовується внутрішній кеш звільнених кортежів малих розмірів (free list до 20 елементів), що усуває накладні витрати на виклики malloc/free при частих операціях створення та видалення.
